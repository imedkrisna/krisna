Bertubi-tubi peristiwa aneh melanda Indoensia. dari [konsultasi MSCI yang mengancam downgrade market Indonesia](https://www.nst.com.my/amp/business/corporate/2026/10/1556569/msci-set-remove-indonesia-global-indices) ke komentar konspirasi yang [menyudutkan Singapura](https://investortrust.id/market/120039/gic-dan-temasek-buka-suara-soal-tuduhan-intervensi-msci-ini-bantahannya), trus tiba-tiba penurunan ekspor gas.

## Ada apa dengan gas?

Kronologinya lucu juga. 21 Oktober Usai Sidang Kabinet Paripurna, setelah Presiden menyebut Indonesia sedang dalam "mode _survival_", Menteri ESDM Bahlil Lahadalia membuka opsi [tidak memperpanjang kontrak ekspor gas ke negara tetangga](https://rmol.id/read/2026/10/21/725801/mode-survival-bahlil-bisa-setop-ekspor-gas-ke-negara-tetangga). Sehari kemudian nadanya naik: Indonesia "tidak segan-segan" mempertimbangkan untuk tidak mengekspor ke negara yang ["tidak menghargai Indonesia"](https://katadata.co.id/berita/energi/6ad9fd5a96a6b/masuk-mode-bertahan-bahlil-ancam-hentikan-ekspor-gas-ke-negara-tak-hargai-ri).

24 Oktober, SKK Migas [mulai mengurangi ekspor gas pipa ke Singapura](https://money.kompas.com/read/2026/10/24/152345126/skk-migas-kurangi-ekspor-gas-ke-singapura-mulai-hari-ini) sebesar 31 BBTUD (miliar BTU per hari): 15 BBTUD dari Natuna dan 16 BBTUD dari Sumatra, yaitu dari WK Jambi Merang yang dikelola PHE. Gas Sumatra itu dialihkan ke pembangkit listrik di Batam, Bintan, dan Sumatra [lewat pipa Transgasindo](https://www.suarasurabaya.net/ekonomibisnis/2026/skk-migas-umumkan-mulai-kurangi-ekspor-gas-ke-singapura/). Besoknya, PM Singapura Lawrence Wong [menjawab](https://fortune.com/2026/10/26/singapore-urges-suppliers-honor-commitments/) bahwa pasokan gas diatur kontrak komersial, dan Singapura "_expect all parties to honor their commitments_". Importir gas mereka akan "mengambil langkah yang diperlukan untuk melindungi hak dan kepentingannya". 

Gongnya, Menteri ESDM [konferensi pers](https://www.youtube.com/watch?v=VNq1e0KgTIs) tanggal 26 Oktober, menyebut bahwa pemotongan 31 BBTUD itu cuma sekitar 8% dari total ekspor ke Singapura yaitu  409,55 BBTUD. Menurutnya angka ini masih di dalam batas kontrak, dan baru [melanggar kalau potongannya 10%](https://jakartaglobe.id/business/just-a-small-cut-indonesia-denies-singapore-gas-contract-breach). [Alasannya dua](https://www.liputan6.com/bisnis/read/8321513/bahlil-pangkas-ekspor-gas-ke-singapura-8-ternyata-ini-alasannya): ada pemeliharaan fasilitas produksi, dan gasnya dibutuhkan PLN serta industri di dalam negeri. 

Sebenernya alasan pemeliharaan mungkin masuk akal ya. potongan 31 BBTUD juga cuma 2% total energi Singapura katanya, so not a huge deal. Tapi yang jadi masalah, di konferensi pers itu Menteri ESDM pasang retorika "[jangan ajari Indonesia soal kontrak](https://wartaekonomi.co.id/read643474/bahlil-sentil-singapura-soal-gas-jangan-ajari-indonesia-soal-kontrak)"[^1]. Ga berenti di situ, Ia juga menyinggung negara tetangga yang ingin dihargai tapi senang mengolok-olok Indonesia, dan meminta kesetaraan serta sikap saling menghormati.

[^1]: menurut indeks [B-Ready](https://www.worldbank.org/en/businessready) milik World Bank, butuh 362 hari untuk menegakkan putusan di Indonesia, sementara Singapura butuh 60 hari. Ga semua yang perlu diajari merasa perlu belajar.

Retorika "sama-sama menghargai" ini sebenarnya problematik. Kontrak gas ke Singapura bukan perjanjian antarnegara. Ia adalah kontrak komersial antara produsen (Medco, PHE, dan para kontraktor Natuna, dengan SKK Migas di sisi negara) dan pembeli seperti Gas Supply Pte Ltd dan Sembcorp. Kewajiban di dalamnya tidak bergantung pada apakah pemerintah Singapura sopan kepada pemerintah Indonesia. Begitu pelaksanaan kontrak dikaitkan dengan "dihargai atau tidak", kontrak komersial berubah jadi alat diplomasi, dan itu justru sinyal yang paling tidak ingin didengar pembeli dan pemodal.

Gw ga belain Singapura sih, cuma sejauh ini sulit untuk bilang bahwa tuduhan di retorika tersebut ga berdasar. Udah gitu, penggunaan alat ekonomi untuk tujuan politik (alias _geoeconomics_) ini juga adalah sinyal yang kurang baik, apalagi ini terjadi setelah MSCI, yang notabene memang banyak pekerjaan rumah dari kitanya juga. Nyalahin pihak eksternal untuk masalah domestik biasanya [punya preseden yang buruk](https://en.wikipedia.org/wiki/Diversionary_foreign_policy).

Wajar kalau pernyataan Pak Bahlil menuai kritik. Salah satunya dari Prof. Mohamad Ikhsan di [Jakarta Post](https://www.thejakartapost.com/opinion/2026/10/26/indonesias-gas-export-cuts-a-costly-blow-to-trust-and-regional-leadership). Menurut beliau, proyek gas hulu dan pipa butuh biaya tetap yang sangat besar dan bersifat _sunk_. Proyek seperti itu bisa dibiayai karena ada pembeli ekspor jangka panjang yang menjadi _anchor demand_. Gas yang dialihkan dari ekspor juga tidak otomatis menemukan pembeli domestik yang sanggup membayar harga setara. Singapura bisa menyesuaikan diri dengan LNG, jadi dampak langsung buat mereka kecil. Ongkos yang sebenarnya ada di reputasi Indonesia sebagai pemasok, dan ini terjadi ketika iklim investasi sudah lesu: survei Kadin mencatat hanya 32,4% perusahaan berencana berinvestasi dalam enam bulan ke depan, turun dari 47,1%.

Di tulisan ini saya mencoba masuk lebih detil soal seberapa penting gas Indonesia buat Singapura, dan sebaliknya? Trus kita cek dari mana gas ke Singapura berasal, siapa yang membangun pipanya, dan kenapa kontrak jangka panjang penting untuk membiayainya (basically ekspresikan argumen prof. ican pake persamaan)? Terakhir, bagaimana kira-kira permainan jangka panjangnya antara Indonesia, Singapura, dan para pemodal gas?

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

latar, ink, abu = "#111827", "#e5e7eb", "#6b7280"
merah, biru, emas = "#ff5c5c", "#22d3ee", "#fbbf24"
hijau, ungu = "#4ade80", "#c084fc"
plt.rcParams.update({"figure.facecolor": latar, "axes.facecolor": latar, "savefig.facecolor": latar,
                     "axes.spines.top": False, "axes.spines.right": False, "axes.grid": False,
                     "axes.edgecolor": ink, "axes.labelcolor": ink, "xtick.color": ink,
                     "ytick.color": ink, "text.color": ink, "legend.labelcolor": ink, "font.size": 10})

MMBTU_PER_TON = 52        # konversi kasar gas alam/LNG; gas pipa ber-CO2 tinggi bisa lebih rendah (~43)

ct = pd.read_csv("comtrade_gas.csv", dtype={"cmdCode": str}).query("partnerCode != 0")
ct["jenis"] = ct.cmdCode.map({"271111": "LNG", "271121": "pipa"})
harga = pd.read_csv("harga_gas.csv", index_col="tahun")
```

## Seberapa penting gas Indonesia buat Singapura?
Gambar 1 menunjukkan impor gas alam Singapura menurut pemasok dan jenisnya, 2014–2025, dalam miliar USD (CIF). Sumbu horizontal adalah tahun. Panel (a) memecah nilai impor menjadi gas pipa dari Indonesia, gas pipa dari Malaysia, dan LNG menurut negara asal. Panel (b) menunjukkan pangsa Indonesia (pipa dan LNG) dalam total impor gas Singapura. Datanya dari UN Comtrade dengan Singapura sebagai pelapor, HS 271111 (LNG) dan 271121 (gas dalam wujud gas, yaitu gas pipa).

```python
sg = ct.query("job == 'sgp_m'").copy()
sg["sumber"] = np.select(
    [(sg.jenis == "pipa") & (sg.partnerDesc == "Indonesia"), (sg.jenis == "pipa") & (sg.partnerDesc == "Malaysia"),
     sg.jenis == "pipa", sg.partnerDesc == "Qatar", sg.partnerDesc == "Australia"],
    ["Pipa Indonesia", "Pipa Malaysia", "Pipa lainnya", "LNG Qatar", "LNG Australia"], "LNG lainnya")
warna = {"Pipa Indonesia": merah, "Pipa Malaysia": emas, "LNG Australia": biru, "LNG Qatar": ungu,
         "LNG lainnya": abu}
g = (sg.groupby(["period", "sumber"]).primaryValue.sum().unstack()
     .reindex(columns=list(warna)).fillna(0) / 1e9)
pangsa_idn = sg.query("partnerDesc == 'Indonesia'").groupby("period").primaryValue.sum() / sg.groupby("period").primaryValue.sum() * 100

fig, ax = plt.subplots(1, 2, figsize=(10, 4.2), gridspec_kw={"width_ratios": [1.6, 1]})
ax[0].stackplot(g.index, g.T.values, colors=list(warna.values()), labels=list(warna), alpha=0.9)
ax[0].set_ylabel("Miliar USD")
ax[0].legend(frameon=False, fontsize=8, loc="upper right", ncol=2)
ax[0].set_title("(a) Impor gas Singapura menurut sumber", loc="left", fontsize=11)
ax[1].plot(pangsa_idn.index, pangsa_idn, color=merah, lw=2)
ax[1].set_ylim(0, 100)
ax[1].set_ylabel("% dari nilai impor gas Singapura")
ax[1].set_title("(b) Pangsa Indonesia", loc="left", fontsize=11)
for a in ax:
    a.set_xlim(2014, 2025)
fig.tight_layout()
plt.show()
```

Ada dua hal yang langsung terlihat. Pertama, Indonesia masih pemasok gas terbesar Singapura: sekitar USD1,7 miliar di 2025, hampir semuanya lewat pipa. Kedua, ketergantungan Singapura pada Indonesia sudah turun drastis. Di 2014 pangsa Indonesia sekitar 65% dari nilai impor gas Singapura, di 2025 tinggal sekitar sepertiga. Penggantinya LNG dari Australia, Qatar, Amerika, dan Mozambik. Ini sejalan dengan [data otoritas Singapura](https://www.mti.gov.sg/newsroom/oral-reply-to-pq-on-impact-of-indonesias-recalibration-of-lng-trade-on-singapores-energy-security): LNG sudah hampir 60% dari pasokan gas mereka, dan sejak 2024 tidak ada satu negara pun yang memasok lebih dari 30%. Maret lalu EMA bahkan sudah bilang kebutuhan energi Singapura tetap bisa dipenuhi dengan LNG walaupun kontrak pipa dari Indonesia dan Malaysia tidak diperpanjang. Jadi ketika EMA menyebut potongan kali ini [kurang dari 2% kebutuhan gas harian mereka](https://ekonomi.republika.co.id/berita/tniamc416/ri-pangkas-ekspor-gas-ke-singapura-bahlil-hanya-dua-persen-dari-kebutuhan-mereka), itu bukan gertakan.

Singapura pinter: mengurangi ketergantungan dari 1 tetangga besar, sesuatu yang Uni Eropa bisa belajar banyak.

Gimana dari sisi Indonesia? Menurut data ekspor Indonesia, Singapura menyerap sekitar 25% nilai ekspor gas alam kita di 2025, di belakang Jepang (28%) dan Tiongkok (27%). Hanya saja, karena gas pipa ke Singapura hanya bisa dijual ke Singapura, kalau Singapura berhenti membeli, gas itu cuma bisa dijual ke tempat yang tersambung pipa ke lapangan yang sama.

Gambar 2 menunjukkan volume ekspor gas pipa Indonesia ke Singapura, 2014–2025, dalam BBTUD. Volumenya dihitung dari berat bersih di data ekspor Indonesia (dari Comtrade) dengan konversi kasar 52 MMBtu per ton. Garis putus-putus menandai angka 409,55 BBTUD yang disebut Bahlil, dan garis titik-titik menandai volume setelah dipotong 31 BBTUD.

```python
pipa = ct.query("job == 'idn_x' & jenis == 'pipa' & partnerDesc == 'Singapore'").groupby("period").netWgt.sum()
bbtud = pipa / 1e3 * MMBTU_PER_TON / 1e3 / 365

fig, ax = plt.subplots(figsize=(8, 4.2))
ax.bar(bbtud.index, bbtud, color=merah, width=0.65)
ax.axhline(409.55, color=ink, lw=1, ls="--")
ax.axhline(409.55 - 31, color=ink, lw=1, ls=":")
lbl = ax.get_yaxis_transform()
ax.text(1.01, 409.55, "409,55 (Bahlil)", transform=lbl, fontsize=8, va="bottom")
ax.text(1.01, 409.55 - 31, "378,55 (setelah\ndipotong)", transform=lbl, fontsize=8, va="top")
ax.set_ylabel("BBTUD (konversi 52 MMBtu/ton)")
ax.set_xticks(bbtud.index)
ax.tick_params(axis="x", labelsize=8)
fig.tight_layout()
plt.show()
# print(bbtud.round(0))
```

Volume ekspor pipa ke Singapura sudah turun kurang lebih separuh dari puncaknya di 2018, dan penurunan ini terjadi jauh sebelum Oktober 2026. Sebagian karena produksi Natuna dan Corridor yang menua. Sebagian lagi karena ketika kontrak Corridor diperpanjang di 2022, volumenya memang [dikecilkan untuk kebutuhan domestik](https://www.thejakartapost.com/business/2022/10/28/indonesia-extends-gas-exports-to-singapore-by-five-years). Dengan konversi 52 MMBtu per ton, volume 2025 sekitar 500 BBTUD. Angka ini sedikit di atas 409,55 BBTUD, wajar karena gas Natuna dan Sumatra mengandung banyak CO₂ sehingga energi per tonnya lebih rendah. Dengan konversi sekitar 43 MMBtu per ton, angkanya cocok.

Apakah ini ilegal? Kalau menurut Men ESDM, potongan 8% belum melanggar kontrak karena memang ada toleransi jika ada masalah teknis dalam _delivery_ sampai 10%. Problemnya, ada ancaman yang ditujukan ke negara yang "tidak menghargai Indonesia" di balik pengurangan itu. Ini problem karena toleransi 10% tadi dibuat dalam itikad baik untuk mengatasi masalah teknis, bukan untuk dipakai membully negara lain.

Berapa ongkos membully Singapura lewat gas? Gambar 3 menunjukkan harga gas, 2014–2025, dalam USD per MMBtu. Garis merah adalah nilai satuan (_unit value_) ekspor gas pipa Indonesia ke Singapura (FOB). Garis biru dan hijau adalah nilai satuan ekspor LNG Indonesia ke Jepang dan Korea Selatan. Ketiganya dihitung dari nilai dibagi berat bersih di data ekspor Indonesia (Comtrade), dikonversi dengan 52 MMBtu per ton. Garis hitam putus-putus adalah harga LNG Jepang dari Pink Sheet Bank Dunia. Pita abu-abu adalah Harga Gas Bumi Tertentu (HGBT) untuk industri dan PLN, USD6 per MMBtu sejak 2020 dan sekitar [USD6,5–7 di 2025–2026](https://www.malaymail.com/news/money/2026/06/29/jakarta-moves-to-cap-lng-prices-as-industry-and-unions-push-back/225662).

```python
idn = ct.query("job == 'idn_x'").groupby(["period", "jenis", "partnerDesc"])[["primaryValue", "netWgt"]].sum()
uv = (idn.primaryValue / (idn.netWgt / 1e3 * MMBTU_PER_TON)).unstack(["jenis", "partnerDesc"])
seri = {("pipa", "Singapore"): ("Pipa ke Singapura", merah), ("LNG", "Japan"): ("LNG ke Jepang", biru),
        ("LNG", "Rep. of Korea"): ("LNG ke Korea Selatan", hijau)}

fig, ax = plt.subplots(figsize=(8, 4.4))
for k, (lab, c) in seri.items():
    ax.plot(uv.index, uv[k], color=c, lw=2, label=lab)
ax.plot(harga.loc[2014:].index, harga.loc[2014:, "lng_jepang"], color=ink, lw=1.2, ls="--", label="LNG Jepang (Pink Sheet)")
ax.fill_between([2020, 2024.5, 2024.5, 2025], [5.9, 5.9, 6.5, 6.5], [6.1, 6.1, 7, 7], color=abu, alpha=0.7,
                lw=0, label="HGBT")
ax.set_xlim(2014, 2025)
ax.set_ylabel("USD per MMBtu")
ax.legend(frameon=False, fontsize=8, loc="upper left", ncol=2)
fig.tight_layout()
plt.show()
#print(pd.DataFrame({lab: uv[k] for k, (lab, _) in seri.items()}).join(harga.lng_jepang).round(2).loc[2019:])
```

Gas pipa ke Singapura dijual dengan harga yang setara atau sedikit di bawah LNG ke Jepang (sekitar USD10 per MMBtu di 2025, dibanding USD10,5), dan di atas LNG ke Korea. Polanya mengikuti harga minyak dengan jeda, karena hampir semua kontrak ekspor gas Indonesia [diindeks ke harga minyak](https://www.ipa.or.id/en/publications/the-gas-business-and-pricing-practices-in-indonesia). Kalau faktor konversinya 43 MMBtu per ton, harganya malah sekitar USD12. Jadi Singapura bukan pembeli murah.

Bandingkan dengan HGBT yang USD6,5–7. Setiap MMBtu yang dialihkan dari Singapura ke pembeli HGBT kehilangan sekitar USD3–3,5. Di skema bagi hasil, selisih harga HGBT ditanggung dari bagian negara. Potongan 31 BBTUD setahun penuh setara sekitar 11 juta MMBtu, jadi penerimaan yang hilang kira-kira USD35–40 juta per tahun. Angka ini tidak besar. Tapi manfaatnya juga tidak besar, dan inilah yang dimaksud Prof. Ikhsan: gas yang dialihkan tidak otomatis mendapat pembeli domestik yang membayar setara. Pembeli domestik mendapat gas lebih murah karena negara memberi subsidi implisit.

Lalu gasnya bisa dialihkan ke mana? Untuk bagian Sumatra, SKK Migas sudah menjelaskan: ke pembangkit di Batam, Bintan, dan Sumatra lewat pipa Transgasindo, yang memang bercabang ke Batam. Untuk bagian Natuna, gasnya cuma bisa pergi ke tempat yang tersambung dengan pipa West Natuna, dan di luar Singapura pilihannya sangat terbatas. Kebutuhan gas industri yang besar ada di Jawa, sedangkan pipa yang menyambungkan Sumatra dan Jawa masih belum lengkap. Gas dari ujung pipa ekspor tidak bisa begitu saja dibawa ke pabrik pupuk di Jawa Timur.

## Dari mana gas ke Singapura berasal, dan siapa yang membiayainya?

Ada dua jalur pipa dari Indonesia ke Singapura. Keduanya dibangun dengan logika yang sama: lapangan gas yang jauh dari pasar domestik baru layak dikembangkan kalau ada pembeli besar yang mau terikat 20 tahun.

| 1999–2000 | Pipa bawah laut West Natuna Transportation System (WNTS), sekitar 656 km dengan kapasitas 700 MMscfd, [dipasang oleh McDermott](https://www.offshore-technology.com/projects/natuna/) |

September 2002, 40% Transgasindo dijual ke Transasia Pipeline (konsorsium yang antara lain berisi ConocoPhillips dan Petronas) [seharga USD187,6 juta](https://www.petromindo.com/news/article/pgn-transasia-pipeline-sign-187-6-million-sale-agreement), termasuk komitmen ekuitas USD57,6 juta |

| 2016 | Medco membeli 40% saham ConocoPhillips di South Natuna Sea Block B |
| 2020 | Pemerintah sempat [memutuskan tidak memperpanjang kontrak Corridor ke Singapura](https://dinsights.katadata.co.id/read/2020/02/20/singapore-gas-allocation-diverted-to-west-java) dan mengalihkannya ke Jawa Barat |

Oktober–November, kontrak Corridor ke GSPL [diperpanjang lima tahun](https://pgjonline.com/news/2022/november/indonesias-medco-energi-signs-5-year-gas-supply-contract-with-singapore) mulai Agustus 2023 dengan volume lebih kecil |
| 2028 | Kontrak-kontrak ekspor gas pipa ke Singapura berakhir |

### Kenapa kontrak jangka panjang penting?

Mari sederhanakan. Sebuah proyek pipa (atau lapangan gas) butuh investasi awal $K$ dan sesudahnya biaya operasi yang kecil. Proyek ini layak dibiayai kalau

$$
  -K + \sum_{t=1}^{T} \frac{(p - c)\, u\, \bar{q}}{(1+r)^t} \geq 0
  \quad \Longleftrightarrow \quad
  p \geq p^* \equiv c + \frac{K \cdot \text{CRF}(r, T)}{u\, \bar{q}}
$$

di mana $p$ adalah harga (atau tarif angkut) per MMBtu, $c$ adalah biaya operasi per MMBtu, $\bar{q}$ adalah kapasitas per tahun, $u$ adalah tingkat utilisasi, yaitu bagian kapasitas yang benar-benar terpakai, $T$ adalah umur proyek, dan $r$ adalah tingkat diskonto, yaitu biaya modal ditambah premi risiko. $\text{CRF}(r,T) = r/[1-(1+r)^{-T}]$ adalah _capital recovery factor_, yang mengubah investasi awal menjadi cicilan tahunan setara. $p^*$ adalah harga impas yang membuat investor _indifferent_ antara invest or not invest.

Ada dua jalur di mana ketidakpastian politik menaikkan $p^*$. Pertama, lewat $u$: kalau pembeli ragu volumenya akan terkirim, atau penjual bisa dipaksa mengalihkan gas, utilisasi yang diharapkan turun. Kedua, lewat $r$: pemberi pinjaman meminta premi risiko yang lebih tinggi. Kontrak jangka panjang dengan klausul _take-or-pay_ (pembeli tetap membayar volume minimum walaupun tidak mengambil gasnya) mengunci $u$ dan menurunkan $r$. Kontrak itulah yang dijadikan jaminan oleh bank. Karena itu kontrak juga harus mengikat penjual. Kalau penjual, atau negara di belakangnya, bisa memotong volume kapan saja, kontrak itu kehilangan nilainya sebagai jaminan. Inilah inti argumen dari prof Ikhsan di oped di atas.

Gambar 4 menunjukkan ilustrasi harga impas untuk komponen modal saja ($p^* - c$), dalam USD per MMBtu, sebagai fungsi utilisasi $u$ (sumbu horizontal). Saya pakai angka rantai Grissik–Singapura: $K$ = USD1,2 miliar (sumur pengembangan ditambah pipa, USD nominal 2001), kapasitas 350 MMscfd (sekitar 360 BBTUD), dan $T$ = 20 tahun. Tiga garis utuh adalah tingkat diskonto 8%, 12%, dan 16%. Garis putus-putus adalah pipanya saja (USD450 juta) pada 8%, sebagai pembanding.

```python
def crf(r, T):
    return r / (1 - (1 + r) ** -T)

K, K_pipa, q_bar, T = 1.2e9, 450e6, 360e3 * 365, 20      # USD, USD, MMBtu per tahun, tahun
u = np.linspace(0.4, 1, 61)

fig, ax = plt.subplots(figsize=(8, 4.2))
for r, c in [(0.08, biru), (0.12, emas), (0.16, merah)]:
    ax.plot(u * 100, K * crf(r, T) / (u * q_bar), color=c, lw=2, label=f"r = {r:.0%}")
ax.plot(u * 100, K_pipa * crf(0.08, T) / (u * q_bar), color=ink, lw=1.2, ls="--", label="pipa saja, r = 8%")
ax.set_ylim(bottom=0)
ax.set_xlabel("Utilisasi yang diharapkan (%)")
ax.set_ylabel("Komponen modal harga impas (USD/MMBtu)")
ax.legend(frameon=False)
fig.tight_layout()
plt.show()
# print(pd.DataFrame({f"r={r:.0%}": K * crf(r, T) / (np.array([1, 0.9, 0.7]) * q_bar) for r in (0.08, 0.12, 0.16)},
#                    index=["u=100%", "u=90%", "u=70%"]).round(2))
```

Dengan $K$ seluruh rantai, komponen modalnya hampir tiga kali lipat dibanding pipa saja: sekitar USD0,9 per MMBtu di utilisasi penuh dan biaya modal 8%. Kalau risiko politik menaikkan biaya modal ke 16% dan utilisasi yang diharapkan turun ke 70%, angkanya jadi sekitar USD2,2, lebih dari dua kali lipat.

Sekilas USD1–2 masih jauh di bawah harga ekspor yang sekitar USD10. Tapi angka ini meremehkan ongkos sebenarnya, setidaknya karena empat hal. Pertama, ini dolar 2001, yang dalam dolar sekarang hampir dua kali lipat. Kedua, belum ada biaya eksplorasi, termasuk sumur-sumur kering yang tidak pernah menghasilkan. Ketiga, belum ada biaya operasi dan penurunan produksi lapangan. Keempat, dan ini yang paling besar, di skema PSC kontraktor tidak menerima seluruh harga: setelah _cost recovery_, sebagian besar bagi hasil gas jatuh ke negara. Margin yang benar-benar dinikmati pemodal jauh lebih tipis daripada selisih USD10 dan USD2, sehingga kenaikan beberapa poin persentase di $r$ bisa jadi pembeda antara proyek jalan dan tidak jalan. Ini makin terasa untuk proyek-proyek baru seperti Andaman, Geng North, dan Abadi yang butuh investasi miliaran dolar.

Yang lebih penting dari angkanya adalah urutan pembiayaannya. Modal untuk kedua rantai ini sebagian besar datang dari luar negeri, dan baru cair setelah ada pembeli Singapura yang terikat _take-or-pay_. Tanpa kontrak dengan Singapura, lapangan Natuna dan Grissik kemungkinan besar tidak dikembangkan secepat itu, dan pipanya tidak akan dibangun. Waktu itu tidak ada pembeli domestik yang tersambung dan sanggup membayar harga yang menutup investasinya.

HGBT menambah satu lapis masalah. Harga gas untuk industri dan PLN dipatok sekitar USD6–7, dan selisihnya terhadap harga kontrak ditanggung dari bagian negara supaya bagian kontraktor tidak berkurang. Jadi kontraktor terlindungi, tapi hanya selama janji itu dipegang dan bagian negara cukup untuk menutupnya. Bagi calon investor eksplorasi, kewajiban memasok gas murah ke domestik berarti satu lagi variabel yang bergantung pada kebijakan, bukan pada pasar. Kalau ditambah preseden bahwa gas ekspor yang sudah dikontrakkan pun bisa dialihkan, harga yang diharapkan turun dan $r$ naik. Dua-duanya membuat eksplorasi dan pengembangan lapangan baru kurang menarik, padahal justru itu yang dibutuhkan untuk memenuhi kebutuhan domestik.

Jadi pemerintah sebenarnya tidak perlu memilih antara gas untuk domestik dan menghormati kontrak ekspor. Gas untuk domestik pun harus dibiayai dulu, oleh pemodal yang sama, yang menghitung $r$ dengan melihat bagaimana negara memperlakukan kontrak-kontrak sebelumnya.

## Permainan jangka panjang

Sekarang kita lihat dari kacamata teori permainan (_game theory_), di mana game yang dimainkan ini _basically prisoner's dilemma_. Ini juga contoh geoekonomi, yaitu memakai alat ekonomi untuk tujuan politik.

Bayangkan permainan kepercayaan (_trust game_) sederhana antara Singapura (S) dan Indonesia (I). Singapura memilih **Percaya**, yaitu meneken atau memperpanjang kontrak jangka panjang dan ikut membiayai infrastrukturnya, atau **Diversifikasi**, yaitu membeli LNG di pasar dan membangun terminal sendiri. Indonesia memilih **Patuh** atau **Potong** (memakai pasokan sebagai alat tawar). Hasilnya ditulis sebagai (hasil Indonesia, hasil Singapura):

| | S: Percaya | S: Diversifikasi |
|---|---|---|
| **I: Patuh** | $(R, R)$ | $(P, P)$ |
| **I: Potong** | $(T, S_0)$ | $(P, P)$ |

dengan _payoff_ $T > R > P > S_0$. $R$ (_reward_) adalah hasil kerja sama: Indonesia mendapat pasar dan pembiayaan, Singapura mendapat gas murah dan stabil. $T$ (_temptation_) adalah hasil Indonesia kalau memotong ketika Singapura sudah terikat. Indonesia mendapat gas untuk domestik plus poin politik, sementara Singapura mendapat $S_0$, hasil terburuk. $P$ adalah hasil kalau Singapura memilih diversifikasi: tidak ada hubungan jangka panjang, dan keduanya berdagang di harga pasar.

Kalau permainan ini cuma sekali, Indonesia punya insentif untuk selalu memilih Potong karena $T > R$. Singapura yang tahu bahwa Indonesia punya _dominant strategy_ ke arah potong itu akan memilih Diversifikasi, dan hasilnya $(P, P)$, lebih buruk bagi keduanya daripada $(R, R)$. Kalau permainan diulang tanpa batas dan Singapura menghukum pemotongan dengan diversifikasi selamanya (_grim trigger_), Indonesia memilih patuh selama

$$
  \frac{R}{1-\delta} \geq T + \frac{\delta P}{1-\delta}
  \quad \Longleftrightarrow \quad
  \delta \geq \delta^* \equiv \frac{T - R}{T - P}
$$

di mana $\delta \in (0,1)$ adalah faktor diskonto Indonesia, yaitu seberapa besar masa depan dihargai. Kerja sama bertahan kalau masa depan cukup penting dibanding godaan sesaat.

Dari kerangka sederhana ini ada tiga hal menarik.

**Pertama, efek akhir permainan.** Kontrak-kontrak pipa ke Singapura berakhir 2028. Di permainan berulang yang ada ujungnya, kerja sama cenderung rontok dari belakang (_backward induction_): di periode terakhir tidak ada masa depan untuk dijaga, jadi Indonesia pilih strategi potong. Singapura mengantisipasi itu dan berdiversifikasi lebih awal alih-alih diversifikasi di belakang. Inilah yang sudah terjadi. Singapura menggeser porsi ke LNG sejak sepuluh tahun lalu (Gambar 1), dan EMA sudah menyatakan tidak bergantung pada perpanjangan kontrak pipa. Dari sisi Singapura, pemotongan ini mengonfirmasi bahwa strategi diversifikasi mereka benar. Responsnya pun bisa ditebak: tidak membalas, menempuh jalur kontraktual lewat importirnya, dan mempercepat diversifikasi (LNG, cadangan minyak di Jurong, bahkan studi nuklir).

**Kedua, permainannya bukan cuma dengan Singapura.** Kalau hanya dengan Singapura, logika akhir permainan bilang memotong menjelang 2028 itu "rasional". Tapi Indonesia bermain dengan banyak pihak sekaligus: pembeli LNG di Jepang, Tiongkok, dan Korea yang menyerap sekitar 70% ekspor gas kita, serta pemodal yang akan membiayai Andaman, Geng North, dan Abadi. Mereka semua mengamati. Dalam model reputasi ala Kreps–Wilson[^2], pemain lain tidak tahu persis "tipe" Indonesia: apakah tipe yang menghormati kontrak (yang sesekali memotong karena alasan teknis), atau tipe yang memakai pemotongan sebagai alat politik. Misalkan $\mu$ adalah keyakinan awal bahwa Indonesia tipe pertama, $\varepsilon$ adalah peluang tipe itu memotong karena gangguan teknis, dan $\sigma$ adalah peluang tipe kedua memotong. Setelah melihat pemotongan, _prior_-nya diperbarui (aturan Bayes) menjadi
$$
  \mu' = \frac{\mu\, \varepsilon}{\mu\, \varepsilon + (1-\mu)\, \sigma}
$$

Kalau pemotongan dijelaskan sebagai pemeliharaan, $\varepsilon$ cukup besar dan $\mu'$ tidak turun banyak. Gangguan teknis memang biasa. Tapi begitu pemotongan dibungkus dengan "negara yang tidak menghargai Indonesia", pengamat tahu ini bukan kejadian teknis. Efektifnya $\varepsilon$ mendekati nol, dan $\mu'$ jatuh. Jadi yang mahal sebenarnya **retorikanya**, bukan 31 BBTUD-nya.

**Ketiga, daya tawar geoekonomi itu bergantung pada ketergantungan yang asimetris.** Sejak Hirschman[^3], kita tahu perdagangan bisa jadi sumber kekuatan politik bagi pihak yang lebih sedikit bergantung. Farrell dan Newman[^4] menyebutnya _weaponized interdependence_. Masalahnya, untuk gas pipa, ketergantungannya justru berat ke sisi kita. Bagi Singapura gas Indonesia tinggal sepertiga impor gas, dan potongan ini kurang dari 2% kebutuhan harian. Bagi Indonesia, Singapura adalah satu-satunya pembeli gas pipa ekspor, dan sebagian gas itu tidak punya pembeli domestik yang realistis. Pipa adalah aset yang spesifik untuk hubungan ini, dan sebagian besar ongkos _sunk_-nya ada di wilayah kita. Ancaman yang paling efektif adalah ancaman yang tidak perlu dieksekusi. Begitu dieksekusi terhadap pihak yang sudah punya alternatif, daya tawarnya habis, dan yang tersisa hanya sinyal buruk ke pemain lain.
